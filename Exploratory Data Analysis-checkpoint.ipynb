## Campaign Count by Platform

```python
platform = (df.groupby('Platform').size().sort_values(ascending=False))

plt.figure(figsize=(7,3))
plt.bar(platform.index,platform.values,color='#3498DB',edgecolor='black')
plt.title('Campaign Count by Platform')
plt.xlabel('platform')
plt.ylabel('counts')
plt.tight_layout()
plt.show()
```

## Revenue Distribution

## ROAS Distribution

```python
plt.figure(figsize=(7,3))

plt.hist(df['ROAS'],bins=20)
plt.title('ROAS Distribution')
plt.xlabel('ROAS')
plt.ylabel('Number of Campaigns')
plt.tight_layout()
plt.show()
```

## Spend vs Revenue

```python
plt.figure(figsize=(7,4))
plt.scatter(df['Spend'],df['Revenue'],alpha=0.6,edgecolor='black')
plt.title('Spend vs Revenue')
plt.xlabel('Spend')
plt.ylabel('Revenue')
plt.tight_layout()
plt.show()
```

## Clicks vs Conversions

```python
plt.figure(figsize=(7,4))
plt.scatter(df['Clicks'],df['Conversions'],alpha=0.6,edgecolor='black')
plt.title('Clicks vs Conversions')
plt.xlabel('Clicks')
plt.ylabel('Conversions')
plt.tight_layout()
plt.show()
```

```python
corr = df[['Spend',
           'Revenue',
           'Clicks',
           'Conversions',
           'CTR',
           'CPC',
           'CPA',
           'ROAS']].corr()
plt.figure(figsize=(7,5))
plt.imshow(corr, cmap='coolwarm', interpolation='nearest')
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns, rotation=45)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

## Outlier Detection

```python
## Revenue

Q1 = df['Revenue'].quantile(0.25)
Q3 = df['Revenue'].quantile(0.75)

IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
revenue_outliers = df[
    (df['Revenue'] < lower) |
    (df['Revenue'] > upper)
]
```

```python
revenue_outliers[
    ['Campaign_ID',
     'Platform',
     'Content_Type',
     'Region',
     'Spend',
     'Revenue',
     'ROAS']
]
```

```python
print("Total Revenue Outliers:", len(revenue_outliers))
```

```python
## ROAS Outliers

Q1 = df['ROAS'].quantile(0.25)
Q3 = df['ROAS'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

roas_outliers = df[
    (df['ROAS'] < lower) |
    (df['ROAS'] > upper)
]

roas_outliers
```

```python
## Spend Outliers

Q1 = df['Spend'].quantile(0.25)
Q3 = df['Spend'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

spend_outliers = df[
    (df['Spend'] < lower) |
    (df['Spend'] > upper)
]

spend_outliers
```

```python
print("Revenue Outliers:", len(revenue_outliers))
print("ROAS Outliers:", len(roas_outliers))
print("Spend Outliers:", len(spend_outliers))
```
